## What is seabron ?

*Seaborn ek Python ki data visualization library hai jo statistical data ko easily samajhne ke liye attractive aur informative graphs banane mein help karti hai.*

```python
import seaborn as sns
sns.get_dataset_names()
```

##### **Seaborn:-** library ko sns naam se import kar rahe hain.
##### **sns.get_dataset_names():-** Ye Seaborn ka function hai jo Seaborn ke saare built-in sample datasets ke naam ki list return karta hai

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

**import seaborn as sns:-** Seaborn library ko import kar rahe hain, jiska use advanced aur attractive graphs banane ke liye hota hai

**import matplotlib.pyplot as plt:-** Matplotlib library ka pyplot module import ho raha hai, jo basic plotting, graph control (title, label, show) ke kaam aata hai.

**import pandas as pd:-** Pandas library ko import kar rahe hain, jo data ko table (DataFrame) form mein handle karne ke liye use hoti hai (CSV read, data cleaning, analysis).

```python
sns.set_theme(style="whitegrid")
```

**sns.set_theme(style="whitegrid"):-** graph ka style set karta hai jisme white background aur grid lines hoti hain, taaki visualization zyada clear ho.

```python
df = sns.load_dataset("tips")
df.head(15)
```

**sns.load_dataset("tips")** Seaborn ka ready-made dataset load karta hai

**df.head()** DataFrame ke first 5 records dikhata hai, taaki data samajhna easy ho.

```python
df.info()
```

**df.info():-** DataFrame ka quick overview deta hai, jisse data ko samajhna aur clean karna easy ho jata hai

```python
df.isnull().sum()
```

**df.info()** DataFrame ka quick overview deta hai, jisse data ko samajhna aur clean karna easy ho jata hai.

```python
df.duplicated().sum()
```

**df.duplicated().sum():-** duplicate rows ki total count nikalne ke liye use hota hai, jo data cleaning mein bahut useful hai.

```python
df.drop_duplicates(inplace=True)
```

**df.drop_duplicates(inplace=True):-** DataFrame se duplicate rows hata deta hai aur change original data mein save kar deta hai.

```python
df.duplicated().sum()
```

```python
df.head(15)
```

```python
# A barplot shows the average (mean) value of a column grouped by categories.

# Example:

# Average tip per day

# Average total bill per gender

# Average price per category

sns.barplot(data=df, x='day', y='tip') # Show average tip per day
plt.show()
```

```python
sns.barplot(data=df, x='sex', y='total_bill') # average total bill per gender
plt.show()
```
