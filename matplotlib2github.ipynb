### *Why are both libraries used together?*

* **NumPy (np):** *Generates data/values.*
* **Matplotlib (plt):** *Plots the graph of that data.*


## Line plot

Create a simple line plot.

```python
x = np.arange(1,10)
y = np.arange(11,20)
plt.plot(x, y, label='Sine Wave')
plt.xlabel('Number upto 1-9')
plt.ylabel('Number upto 11-19')
plt.title('Simple Line Plot')
plt.grid()
plt.legend()
plt.show()
```

* **x variable:** A NumPy array is created with values starting from 1 up to 9 (10 is not included).
* **y variable:** A NumPy array is created with values starting from 11 up to 19.
* **plt.plot(x, y, label="Sine Wave"):** Creates a simple line plot of x and y values and assigns a label to the line.
* **plt.xlabel("Number upto 1-9"):** Sets the label for the x-axis.
* **plt.ylabel("Number upto 11-19"):** Sets the label for the y-axis.
* **plt.title("Simple Line Plot"):** Sets the title of the graph.
* **plt.grid():** Displays a grid in the background of the graph.
* **plt.legend():** Shows the label ("Sine Wave") on the graph.
* **plt.show():** Displays the final line plot on the screen.


## scatter plot

To understand the relationship between two variables, their values are visually displayed using a graph.


```python
x = np.random.rand(50)
y = np.random.rand(50)
plt.scatter(x, y, color='b' , marker='*')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot')
plt.show()
```

* **x variable:** Uses NumPy to generate 20 random numbers between 0 and 1.
* **y variable:** Uses NumPy to generate 20 random numbers between 0 and 1.
* **plt.scatter(x, y, color="blue", marker="o"):** Creates a scatter plot of x and y values with blue color and "o" marker.


## Bar chart

A bar graph is used to show the relationship between alphabets and numbers. In this graph, each alphabet is represented by a rectangular bar corresponding to its number.


```python
categories = ['A', 'B', 'C', 'D' ,'E']
values = [3, 7, 1, 8 ,5.8]
plt.bar(categories, values, color='blue')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Chart')
plt.show()
```

* **categories variable:** Stores a list of category names.
* **values variable:** Stores a list of numerical values.
* **plt.bar(categories, values, color="green"):** Creates a bar chart where categories are on the x-axis, values determine the height of the bars, and the bars are colored green.


## Histogram

A histogram shows the distribution of numerical data, where the data is divided into ranges (bins) and their frequencies are represented using bars.


```python
data = np.random.randn(1000)
plt.hist(data, bins=70, color='red', edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram')
plt.show()
```

* **plt.hist(data, bins=70, color='red', edgecolor='black'):** Creates a histogram with 70 bins, red-colored bars, and black edges.
* **plt.xlabel('Value'):** Labels the x-axis as “Value.”
* **plt.ylabel('Frequency'):** Labels the y-axis as “Frequency.”


# sub plot


```python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].plot(x, y, color='blue', linestyle='--')
ax[0].set_title('Sine Wave')
ax[1].bar(categories, values, color='red')
ax[1].set_title('Bar Chart')
plt.show()
```

* **fig, ax = plt.subplots(1, 2, figsize=(10, 4)):** Creates a figure with 2 side-by-side subplots, with a size of 10×4 inches.
* **ax[0].plot(x, y, color='blue', linestyle='--'):** Plots a dashed blue line of x and y values in the first subplot.
* **ax[0].set_title('Sine Wave'):** Sets the title of the first subplot as “Sine Wave.”
* **ax[1].bar(categories, values, color='red'):** Creates a red-colored bar chart of categories and values in the second subplot.
* **ax[1].set_title('Bar Chart'):** Sets the title of the second subplot as “Bar Chart.”
* **plt.show():** Displays both subplots together on the screen.
